```python
import datetime
```

```python
dt=datetime.datetime.now()
dt
```

```python
data['Age']=dt.year - data['year']
data
```

```python
print(data['fuel_type'].unique())
```

```python
data['fuel_type']=data['fuel_type'].map({'Diesel':0,'Petrol':1,'CNG':2})
data
```

```python
print(data['seller_type'].unique())
```

```python
data['seller_type']=data['seller_type'].map({'Individual':0,'Dealer':1,'Trustmark Dealer':2})
data
```

```python
print(data['transmission'].unique())
```

```python
data['transmission']=data['transmission'].map({'Manual':0,'Automatic':1})
data
```

```python
x = data.drop(['year', 'selling_price'], axis=1)
```

```python
y = data['selling_price']
y
```

```python
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
rf = RandomForestRegressor()
rf.fit(x_train, y_train)
```

```python
gb = GradientBoostingRegressor()
gb.fit(x_train, y_train)
```

```python
xgb = XGBRegressor()
xgb.fit(x_train, y_train)
```

```python
y_pred1 = lr.predict(x_test)
y_pred2 = rf.predict(x_test)
y_pred3 = gb.predict(x_test)
y_pred4= xgb.predict(x_test)
```

```python
from sklearn import metrics
```

```python
score1 = metrics.r2_score(y_test, y_pred1)
score2 = metrics.r2_score(y_test, y_pred2)
score3 = metrics.r2_score(y_test, y_pred3)
score4 = metrics.r2_score(y_test, y_pred4)
```

```python
score1, score2, score3, score4
```

```python
gb_final = GradientBoostingRegressor()
gb_final.fit(x,y)
```

```python
import matplotlib.pyplot as plt

algorithms = {
    "Linear Regression": 0.9419685091243681,
    "Random Forest Regression": 0.9207337619552343,
    "Gradient Boosting Regression": 0.9309594431058055,
    "XGBoost Regression": 0.902609646320343
}

plt.figure(figsize=(10, 6))

plt.bar(algorithms.keys(), algorithms.values(), color=["blue", "green", "orange", "red"])

plt.xlabel("Algorithms")
plt.ylabel("Accuracy")
plt.title("Algorithm Accuracy Comparison")
plt.ylim(0, 1)

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

scores = [
    0.9419685091243681,
    0.9207337619552343,
    0.9309594431058055,
    0.902609646320343
]

labels = [
    "Linear Regression",
    "Random Forest Regression",
    "Gradient Boosting Regression",
    "XGBoost Regression"
]

plt.plot(labels, scores, marker='o', linewidth=2)

plt.xlabel("Algorithms")
plt.ylabel("Accuracy")
plt.title("Algorithm Accuracy Comparison - Line Graph")
plt.ylim(0, 1)
plt.grid(True, linestyle='--', alpha=0.5)

# Values on points
for i, value in enumerate(scores):
    plt.text(i, value + 0.02, f"{value:.4f}", ha='center')

plt.xticks(rotation=15)
plt.tight_layout()

plt.show()
```
